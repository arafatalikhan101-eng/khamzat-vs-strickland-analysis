```python
#importing libraries
import pandas as pd
import matplotlib.pyplot as plt
```

```python
#creating data based on available ufc information
data = [
    ["Khamzat Chimaev", 13, 0, 6, 5, 2, 7.5, 2.2, 3.5],
    ["Sean Strickland", 28, 5, 11, 4, 13, 5.9, 4.0, 0.8]
]

columns = [
    "Fighter", "Wins", "Losses", "KO Wins", "Sub Wins",
    "Decision Wins", "Strikes/min", "Strikes Absorbed", "Takedowns Avg"
]

df = pd.DataFrame(data, columns=columns)

df
```

```python
#chimaev's win breakdown
chimaev = df[df["Fighter"] == "Khamzat Chimaev"]

labels = ["KO", "Submission", "Decision"]
values = [
    chimaev["KO Wins"].values[0],
    chimaev["Sub Wins"].values[0],
    chimaev["Decision Wins"].values[0]
]

plt.figure()
plt.pie(values, labels=labels, autopct="%1.1f%%")
plt.title("Khamzat Chimaev - Win Breakdown")
plt.show()
```

```python
#striking output graph
plt.figure(figsize=(6,4))

bars = plt.bar(df["Fighter"], df["Strikes/min"])

bars[0].set_color("#e50914")
bars[1].set_color("#444444")

plt.gca().set_facecolor("black")
plt.gcf().patch.set_facecolor("black")

for spine in plt.gca().spines.values():
    spine.set_visible(False)

plt.xticks(color="white")
plt.yticks(color="white")

plt.ylabel("Strikes/min", color="white")

for i, v in enumerate(df["Strikes/min"]):
    plt.text(i, v + 0.2, str(v), color="white", ha='center')

plt.tight_layout()
plt.show()
```

```python
#grappling control
plt.figure(figsize=(6,4))

bars = plt.bar(df["Fighter"], df["Takedowns Avg"])

bars[0].set_color("#e50914")
bars[1].set_color("#444444")

plt.gca().set_facecolor("black")
plt.gcf().patch.set_facecolor("black")

for spine in plt.gca().spines.values():
    spine.set_visible(False)

plt.xticks(color="white")
plt.yticks(color="white")

plt.ylabel("Takedowns", color="white")

for i, v in enumerate(df["Takedowns Avg"]):
    plt.text(i, v + 0.1, str(v), color="white", ha='center')

plt.tight_layout()
plt.show()
```

```python
#insights
print("Key Insights:\n")

if df.loc[0, "Takedowns Avg"] > df.loc[1, "Takedowns Avg"]:
    print("Chimaev has a strong grappling advantage.")

if df.loc[0, "Strikes/min"] > df.loc[1, "Strikes/min"]:
    print("Chimaev is more explosive in striking.")

if df.loc[1, "Strikes Absorbed"] > df.loc[0, "Strikes Absorbed"]:
    print("Strickland absorbs more strikes but shows durability.")
```
